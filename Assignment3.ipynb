```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
import seaborn as sns
```

```python
1. Countplot: Passenger Count by Embarked Port
● Create a countplot to display the number of passengers who boarded from each
port (embarked), separated by sex.
```

```python

titanic = sns.load_dataset("titanic")
plt.figure(figsize=(8, 6))
sns.countplot(data=titanic, x="embarked", hue="sex", palette="Set2")
plt.title("Passenger Count by Embarked Port (Split by Sex)", fontsize=14, fontweight="bold")
plt.xlabel("Port of Embarkation (C = Cherbourg, Q = Queenstown, S = Southampton)")
plt.ylabel("Number of Passengers")
plt.legend(title="Sex")
plt.tight_layout()
plt.savefig("1_countplot_embarked.png", dpi=150)
plt.show()
```

2. Swarmplot: Age Distribution across Classes
● Create a swarmplot to visualize the distribution of age across passenger classes
(pclass), categorized by sex.

```python


titanic = sns.load_dataset("titanic")

plt.figure(figsize=(9, 6))
sns.swarmplot(data=titanic, x="pclass", y="age", hue="sex" ,dodge=True, size=3, palette="Set1")
plt.title("Age Distribution Across Passenger Classes (by Sex)", fontsize=14, fontweight="bold")
plt.xlabel("Passenger Class")
plt.ylabel("Age (years)")
plt.legend(title="Sex")
plt.tight_layout()

plt.show()

# Observation:
# Median age rises with class: 1st class skews older (30s-40s), 3rd class is
# younger and denser, reflecting immigrant/working-class families. Within each
# class, male and female age spreads look similar.
```

5. Violin Plot: Fare by Class and Gender
● Create a violin plot to visualize fare distribution across classes, separated by
gender.



```python

sns.set_theme(style="whitegrid")
titanic = sns.load_dataset("titanic")

plt.figure(figsize=(9, 6))
sns.violinplot(data=titanic, x="pclass", y="fare", hue="sex", split=True, palette="Pastel1")
plt.title("Fare Distribution by Class and Gender", fontsize=14, fontweight="bold")
plt.xlabel("Passenger Class")
plt.ylabel("Fare ($)")
plt.legend(title="Sex")
plt.tight_layout()
plt.show()

# Observation:
# Fare distributions are heavily right-skewed in every class, with 1st class
# showing by far the widest spread and highest outliers. Within each class,
# female fares tend to sit slightly higher than male fares, especially in 1st class.
```
