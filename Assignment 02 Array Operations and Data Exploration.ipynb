4. Write a NumPy program that creates a 2D array x of shape (3, 4) and a 1D array y of
shape (4,). Subtract y from each row of x using broadcasting. (Score: 2)

```python
print(f"This is {x.ndim}D array")
x=np.array([[1,2,3,4],[5,6,7,8],[8,3,4,5]])
print(x)

print(f"This is {y.ndim}D array")
y = np.array([2,2,2,2])
print(y)
print("Result")
result= x - y
print(result)
```

5. Create a dataframe with the following columns: name, age, and gender. The dataframe
should have 10 rows of data. (Score: 2)

```python
df = pd.DataFrame({
    "name":["pranav","umer","adhil","shone","paraveen","said","jithin","daine","anju","aswathy"],
    "age" :[27,31,24,25,26,24,28,26,28,32],
    "gender":["M","M","M","M","M","M","M","F","F","F"]
})
print(df)
```

5.1) Add a new column to the data frame created in question 1, called occupation.
The values for this column should be Programmer, Manager, and Analyst,
corresponding to the rows in the dataframe. (Score: 1)

```python
ddf=[' Programmer',' Manager', 'Analyst,']
df['occupation']=np.resize(ddf,len(df))
df
```

5.2) Select the rows of the dataframe where the age is greater than or equal to 30.
